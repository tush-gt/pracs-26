```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from sklearn.metrics import mean_absolute_error, mean_squared_error

experience = np.array([1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.2, 3.7])
salary = np.array([39.3, 46.2, 37.7, 43.5, 39.8, 56.6, 60.1, 54.4, 64.4, 57.1])

slope, intercept, r_value, p_value, std_err = stats.linregress(
    experience, salary
)

salary_pred = slope * experience + intercept

r2 = r_value**2
mse = mean_squared_error(salary, salary_pred)
mae = mean_absolute_error(salary, salary_pred)

print(f"Slope (m): {slope:.4f}")
print(f"Intercept (c): {intercept:.4f}")
print(f"R-squared (Accuracy): {r2:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")

plt.figure(figsize=(8, 5))
plt.scatter(experience, salary, color="blue", label="Actual Data Points")
plt.plot(
    experience, salary_pred, color="red", label=f"Fit Line (R² = {r2:.2f})"
)
plt.title("Linear Regression Analysis")
plt.xlabel("Years of Experience")
plt.ylabel("Salary ($k)")
plt.legend()
plt.grid(True)
plt.show()
```
